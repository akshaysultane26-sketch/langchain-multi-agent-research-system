```python
from langchain_core.tools import tool

@tool
def get_weather_data(city: str) -> str:
    """Fetch current weather information for a city."""
    url = (
        f"https://api.weatherstack.com/current?"
        f"access_key={os.getenv('WEATHERSTACK_API_KEY')}&query={city}"
    )
    response = requests.get(url)
    data = response.json()

    if "current" not in data:
        return f"Could not fetch weather data for {city}"

    return f"{data['current']['temperature']}°C, {data['current']['weather_descriptions'][0]}"
```

```python
print(get_weather_data.invoke("Pune"))
```

```python
import requests
from langchain_core.tools import tool

@tool
def get_weather_data(city: str) -> str:
    """Fetch current weather information for a city."""
    url = (
        f"https://api.weatherstack.com/current?"
        f"access_key={os.getenv('WEATHERSTACK_API_KEY')}&query={city}"
    )
    response = requests.get(url)
    data = response.json()

    if "current" not in data:
        return f"Could not fetch weather data for {city}"

    return f"{data['current']['temperature']}°C, {data['current']['weather_descriptions'][0]}"
```

```python
tools = [search_tool, get_weather_data]
```

```python
agent = create_agent(
    llm,
    tools,
    system_prompt=(
        "You are a helpful assistant. "
        "For weather-related questions, always use the get_weather_data tool. "
        "For other current-events or general questions, use the search tool."
    ),
)
```

```python
result = agent.invoke(
    {"messages": [HumanMessage("What's the weather in Pune right now?")]}
)
for msg in result["messages"]:
    print(type(msg).__name__, "-", getattr(msg, "content", msg))
    print("---")
```
